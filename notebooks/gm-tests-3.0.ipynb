### 1.1 Temporal Kz Convention

GM is written in a **temporal Kz convention**. The speed scale is normalised by
$c$, so transport speeds are handled through the dimensionless ratio
$$\beta = \frac{v}{c}, \qquad 0 \le \beta \le 1.$$
When a transport speed is written as $1$, this means **one light-speed unit**
in the normalised ratio, not that Planck's constant has been removed from the
theory.

Planck's constant is **not** set to $1$. Instead, energy-like quantities are
represented as temporal rates by division by $h$. In particular,
$$E_{\rm Kz} = \frac{E}{h}, \qquad
  m_{\rm Kz} = \frac{m c^2}{h}, \qquad
  T_{\rm Kz} = \frac{k_B T}{h}, \qquad
  P_{\rm Kz} = \frac{P}{h}.$$
Thus a catalog mass in kilograms is read numerically in **Kz / Hz** through its
rest-energy rate:
$$f_{\rm Kz}(M) \equiv m_{\rm Kz}(M) = \frac{M c^2}{h}.$$
All core GM relations below are therefore written in temporal Kz variables such
as $\mu$, $p$, $b$, $R$, $f_e$, and $\alpha$, with $h$ kept explicit in the
dictionary layer.

### 1.2 Numerical Dictionary (Sunit + SI-2018)

For numerical comparisons we embed GM into SI-2018 using a dictionary:

- **Sunit mapping** for catalog masses $M$ (kg):
  $$f_{\rm Kz}(M) = \frac{M c_{\rm SI}^2}{h_{\rm SI}}.$$  
- **Numerical vacuum map** from Kz frequency to a length scale:
  $$\mu_{\rm num} = \kappa_{\rm vac}^{(\rm num)} f_{\rm Kz}(M),\qquad
    \kappa_{\rm vac}^{(\rm num)} = \frac{G_{\rm SI} h_{\rm SI}}{c_{\rm SI}^4}.$$

These expressions belong to the **dictionary layer**, not to GM itself. They provide numerical values for the Kz temporal rates and for $\mu$ in SI units. GM corollaries below remain written in temporal Kz variables.

### 1.3 Small-Gradient Linearisation (Nonlinear → Linear)

In the Solar tests we work in a **small-gradient regime**

```python
# 1. INPUTS (SI-2018 NUMERICAL LAYER)
# ===================================

import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

# Exact constants (SI 2018)
c_SI = 299_792_458.0      # m/s
h_SI = 6.62607015e-34     # J s

# Measured constants (CODATA)
G_SI = 6.67430e-11        # m^3 kg^-1 s^-2
alpha_fs = 7.297352569e-3 # Fine structure constant

# Solar system ephemerides (catalog)
M_sun_kg = 1.98847e30
R_sun_m  = 695_700.0 * 1000.0
a_mer_m  = 57_909_050.0 * 1000.0
e_mer    = 0.205630
T_mer_s  = 87.9691 * 86400.0
century_s = 36525.0 * 86400.0   # seconds per Julian century

# Hydrogen parameters
m_e_kg = 9.10938356e-31
a0_ref = 5.291772109e-11        # Bohr radius (m, NIST)

# Angular conversion: radians -> arcseconds
ARCSEC = 3600.0 * 180.0 / math.pi

print("SI-2018 numerical environment initialised.")
```

```python
# 2. SUNIT MAPPING AND NUMERICAL mu
# =================================
# Temporal Kz convention: transport speed is normalised by c,
# while energy-like quantities are represented as rates via division by h.
# This cell provides SI numerical values for those Kz rates and for mu.

def mass_to_Kz(mass_kg: float) -> float:
    """Dictionary: catalog mass [kg] -> Kz temporal rate [Hz] via m_z = m c^2 / h."""
    return (mass_kg * c_SI**2) / h_SI

# Kz temporal rates (numerical)
f_sun_hz = mass_to_Kz(M_sun_kg)
f_elec_hz = mass_to_Kz(m_e_kg)

# Numerical GM length scale: mu_num = kappa_vac_num * f_sun
KAPPA_VAC_NUM = (G_SI * h_SI) / (c_SI**4)
mu_m = KAPPA_VAC_NUM * f_sun_hz

print(f"Solar Kz temporal rate (Sunit): f_sun = {f_sun_hz:.4e} Hz")
print(f"Electron Kz temporal rate    : f_e   = {f_elec_hz:.4e} Hz")
print(f"Numerical GM source length  : mu    = {mu_m:.4f} m")
```

### Corollary I — Orbital Precession (GM annulus)

**Setup.** 
Closed probe motif (Mercury) in the refractive gradient of a central Kz source
with GM length scale $\mu$. The channel is in a small-gradient regime
$$\epsilon(r) = \left|\frac{\mu}{r}\right| \ll 1,$$
so first-order terms in $\mu/p$ dominate.

**GM consequence.** 
From GM Universal Wave Transport and Nonlinear Refraction, the refractive
gradient induces a cubic correction in the orbital closure. In the linearised
regime the net trajectory shift per revolution satisfies
$$\Delta \phi_{\text{orbit}} = \frac{6\pi\,\mu}{p},$$
where $p$ is the semilatus rectum of the motif. This expression is obtained
from GM annulus closure in Kz units; no external field equations are used as
premises.

**Specialisation (Sun–Mercury).** 
For Mercury,
$$p_{\text{mer}} = a_{\text{mer}} (1 - e_{\text{mer}}^2),$$
with catalog semimajor axis $a_{\text{mer}}$ and eccentricity $e_{\text{mer}}$.
The per-orbit shift is
$$\Delta \phi_{\odot\text{–mer}} = \frac{6\pi\,\mu}{p_{\text{mer}}},$$
and the shift per century multiplies this by the number of Mercury orbits per
century.

**Calibration.** 
The numerical value of $\mu$ for the Sun is supplied by the dictionary cell
above. The GM expression itself remains in Kz form.

**Falsification.** 
If the residual between the GM prediction and the catalog residual advance
satisfies
$$\bigl|\Delta\phi_{\rm GM} - \Delta\phi_{\rm obs}\bigr| > 0.1\,\text{arcsec/century},$$
after accounting for catalog Newtonian contributions, then this corollary is
falsified in the Sun–Mercury window.

**Provenance.** GM §3.3 (Nonlinear refraction), GM §4.6.1 (Geometry as tool), AF §2 (mapping and calibration).


```python
# TEST I: MERCURY PERIHELION
# ==========================

# Semilatus rectum
p_mer = a_mer_m * (1.0 - e_mer**2)

# GM shift per orbit (radians)
shift_rad = (6.0 * math.pi * mu_m) / p_mer

# Arcsec per century
arcsec_cy = shift_rad * ARCSEC * (century_s / T_mer_s)
obs_cy = 42.98  # catalog residual (arcsec/century)
rel_err_peri = 100.0 * (arcsec_cy - obs_cy) / obs_cy

print("[Test I] Orbital Precession (Sun–Mercury)")
print(f"  GM prediction     : {arcsec_cy:.4f} arcsec/cy")
print(f"  Catalog reference : {obs_cy:.4f} arcsec/cy")
print(f"  Rel. error        : {rel_err_peri:.3e} %")

# Visualisation: accumulation over 100 years
years = np.linspace(0.0, 100.0, 200)
y_gm = (arcsec_cy / 100.0) * years

plt.figure(figsize=(8, 4))
plt.plot(years, y_gm, label="GM prediction", color='b')
plt.scatter([100], [obs_cy], color='r', label="Catalog (100 y)", zorder=5)
plt.title("Corollary I: Mercury precession accumulation")
plt.xlabel("Time (years)")
plt.ylabel("Shift (arcsec)")
plt.legend(); plt.grid(True, linestyle=':')
plt.tight_layout()
plt.show()
```

#### Corollary IIb — Entropic Shape Test ($1/b^2$ vs $1/b$)

**Setup.** 
GM allows matter/noise corrections to the effective index, modelling capacity
limits in the channel. An entropic variant with a $1/r^2$ correction yields a

$$\delta_{\rm ent}(b) = \frac{4\mu R_\odot}{b^2},$$

**Shape comparison.**

$$\delta_{\rm GM}(b) = \frac{4\mu}{b}, \qquad
  \delta_{\rm ent}(b) = \frac{4\mu R_\odot}{b^2}.$$
The residual
$$\Delta\delta(b) = \delta_{\rm ent}(b) - \delta_{\rm GM}(b)$$
is a GM prediction for off-limb deflection shape.

**Provenance.** GM §3.3 (Nonlinear refraction with matter/noise), GM §4.7 (Confinement motifs).


```python
# TEST II & IIb: LIGHT DEFLECTION (LIMB + SHAPE)
# ==============================================

b_limb = R_sun_m
defl_rad_gm = (4.0 * mu_m) / b_limb
defl_arcsec_gm = defl_rad_gm * ARCSEC
defl_arcsec_baseline = (2.0 * mu_m / b_limb) * ARCSEC
obs_lens = 1.75

rel_err_limb = 100.0 * (defl_arcsec_gm - obs_lens) / obs_lens

print("\n[Test II] Light deflection at Solar limb")
print(f"  GM prediction     : {defl_arcsec_gm:.4f} arcsec")
print(f"  Catalog reference : {obs_lens:.4f} arcsec")
print(f"  Rel. error        : {rel_err_limb:.3e} %")
print(f"  Baseline (2 mu/b) : {defl_arcsec_baseline:.4f} arcsec (reference)")

b_vals_shape = np.linspace(1.0, 5.0, 200)  # in Solar radii
b_m_shape = b_vals_shape * R_sun_m
defl_geom = (4.0 * mu_m / b_m_shape) * ARCSEC
defl_ent = (4.0 * mu_m * R_sun_m / (b_m_shape**2)) * ARCSEC
residual_shape = defl_ent - defl_geom

fig = plt.figure(figsize=(10, 7))
gs = gridspec.GridSpec(2, 1, height_ratios=[3, 1])

ax0 = plt.subplot(gs[0])
ax0.plot(b_vals_shape, defl_ent, label=r"Entropic GM ($\propto 1/b^2$)", linewidth=2.0)
ax0.plot(b_vals_shape, defl_geom, "--", label=r"Geometric GM ($\propto 1/b$)", alpha=0.8)
ax0.axvline(1.0, color='r', linestyle=':', label='Solar limb')
ax0.set_ylabel("Deflection (arcsec)")
ax0.set_title("Corollary II & IIb: Light deflection shape (GM variants)")
ax0.legend(loc="upper right")
ax0.grid(True, alpha=0.3)

ax1 = plt.subplot(gs[1], sharex=ax0)
ax1.plot(b_vals_shape, residual_shape)
ax1.axhline(0.0, color="k", linewidth=1.0)
ax1.set_xlabel("Impact parameter (Solar radii)")
ax1.set_ylabel("Entropic − geometric\n(arcsec)")
ax1.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# TEST III: SOLAR SURFACE REDSHIFT
# =================================

z_surf = mu_m / R_sun_m
obs_z = 2.12e-6
rel_err_z = 100.0 * (z_surf - obs_z) / obs_z

print("\n[Test III] Solar surface redshift")
print(f"  GM prediction  : {z_surf:.6e}")
print(f"  Catalog value  : {obs_z:.6e}")
print(f"  Abs. error     : {z_surf - obs_z:.3e}")
print(f"  Rel. error     : {rel_err_z:.3e} %")

r_vals = np.linspace(R_sun_m, 10.0 * R_sun_m, 200)
z_curve = mu_m / r_vals

plt.figure(figsize=(8, 4))
plt.plot(r_vals / R_sun_m, z_curve, label="GM prediction", color='g')
plt.scatter([1.0], [obs_z], color='r', label="Surface anchor", zorder=5)
plt.title("Corollary III: Surface redshift profile (Solar window)")
plt.xlabel("Radius (Solar radii)")
plt.ylabel("Redshift z")
plt.legend(); plt.grid(True, linestyle=':')
plt.tight_layout()
plt.show()
```

```python
# TEST V: HYDROGEN SHELL RADII (GM vs Bohr)
# =========================================

def calc_hydrogen_radius(n, f_e=f_elec_hz, alpha=alpha_fs):
    """GM Kz radius: r_n = (n^2 * c) / (2*pi * f_e * alpha) in numerical SI."""
    return (n**2 * c_SI) / (2.0 * math.pi * f_e * alpha)

n_levels = np.array([1, 2, 3, 4])
radii_gm = np.array([calc_hydrogen_radius(n) for n in n_levels])
radii_ref = a0_ref * n_levels**2
residuals = radii_gm - radii_ref
rel_err = 100.0 * residuals / radii_ref

df_h = pd.DataFrame({
    "n": n_levels,
    "GM radius r_n (m)": radii_gm,
    "Ref radius n^2 a0 (m)": radii_ref,
    "Abs. error (m)": residuals,
    "Rel. error (%)": rel_err,
})

pd.set_option('display.float_format', '{:.6e}'.format)
print("\n[Test V] Hydrogen shells (GM vs Bohr)")
display(df_h)
```

### Corollary VI — Galactic Rotation Curves (GM update and GR comparison)

**Axiom A1 (Dynamic Metric Units).** Under the SI-2018 standard, the base units of length ($u_m$) and mass ($u_{kg}$) are explicitly derived from the local unit of time ($u_s$). In a non-uniform time-dilation field, $u_s$ varies with position. Therefore, $u_m$ and $u_{kg}$ are dynamically time-dependent variables: $u_m(t)$, $u_{kg}(t)$.

**Choice D1 (Temporal Mapping).** All observables are written strictly in temporal/informational rates to prevent chain-rule omissions from composite units. The macroscopic conversion factor $G$ is bypassed at the derivation stage.

**Methodological note.** In the AFC reading used here, calculus is **not treated as primitive continuum bookkeeping**. The primitive layer is discrete and ternary; the continuum expressions below are the SI comparison layer attached to that discrete structure.

**Consequence C1 (Rigorous Weak-Field).** Evaluating the physical radial acceleration of an orbiting body requires differentiating the position-unit product: $\frac{d^2}{dt^2}[r(t) u_m(t)]$. From A1 by the product rule, this yields an acceleration of $\ddot{r}u_m + 2\dot{r}\dot{u}_m + r\ddot{u}_m$.

**Consequence C2 (GM 2D Confinement).** In the GM framework, absolute grids are discarded. The galactic disc acts as a 2D confinement window. By GM tube invariance, the boundary area scales as $A_2(r) \propto r$, forcing the frame amplitude to scale as $|E_I| \propto 1/r$.

**Corollary VI (Flat Rotation Residue).** For a probe at radius $r \gg R_{\text{core}}$, the orbital velocity flattens to a constant $v = \text{const}$.

*Derivation.* From C1, if the unit acceleration term $\ddot{u}_m$ is driven by a spatial gradient scaling as $1/r^2$, then the geometric contribution $r\ddot{u}_m$ yields a constant $v^2$ residue. From C2, the GM 2D confinement window with $A_2(r) \propto r$ gives a frame amplitude $|E_I| \propto 1/r$, which is the scaling associated with a flat orbital-speed profile in this comparison window. The GR-style temporal-unit calculation and the GM confinement scaling therefore meet on the same flat-curve residue without introducing an additional mass sector inside the notebook.

**Falsification.** If flat galactic rotation curves are shown to require a kinematic profile that diverges from a constant $v^2$ residue outside the core radius, this corollary is challenged.

**Provenance.** GM §4.6.1 (Geometry as tool), GM §4.6.2 (Dimensional flows & confinement), AFC discrete Stokes bookkeeping.


```python
# TEST VI: GALACTIC ROTATION (GM UPDATE + GR COMPARISON)
# =====================================================

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

# -- 1. Temporal Inputs (No 'G' or composite units) --
# We use c_SI from the previously initialized SI-2018 environment
v_catalog_km_s = 220.0
v_target_dim = (v_catalog_km_s * 1000.0) / c_SI

# Define Galactic Bulge (3D) to Disc (2D) boundary (~2.0 kpc for Milky Way)
R_core_kpc = 2.0
radii_kpc = np.linspace(2.0, 24.0, 100)

# -- 2. Model A: Classical / Keplerian baseline --
# Static-unit proxy: drops as 1/sqrt(r), missing the flat target.
v_classical_dim = v_target_dim * np.sqrt(R_core_kpc / radii_kpc)
v_classical_km_s = (v_classical_dim * c_SI) / 1000.0

# -- 3. Model B: Rigorous GR-style temporal mapping (Chain Rule Applied) --
# Keeps the r*ddot{u}_m geometric unit-derivative term (Consequence C1).
# ddot{u}_m scales as 1/r^2, integrating to a constant v^2 residue.
v_geom_dim_sq = v_target_dim**2
v_fixed_dim = np.sqrt(v_classical_dim**2 + v_geom_dim_sq * (1.0 - R_core_kpc / radii_kpc))
v_fixed_km_s = (v_fixed_dim * c_SI) / 1000.0

# -- 4. Model C: General Mechanics (2D confinement flow) --
# Pure 2D disc boundary area A_2(r) ~ r, leading to E_I ~ 1/r (Consequence C2).
v_gm_dim = np.full_like(radii_kpc, v_target_dim)
v_gm_km_s = (v_gm_dim * c_SI) / 1000.0

# -- 5. Visualisation --
fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111)

# Observational Data Anchor
ax.scatter(
    [4, 8, 12, 16, 20, 24],
    np.full(6, v_catalog_km_s),
    color='k',
    marker='s',
    s=40,
    zorder=5,
    label='Catalog target (flat)',
)

# Plot Models
ax.plot(
    radii_kpc,
    v_classical_km_s,
    'r--',
    linewidth=2,
    label='Classical / Keplerian baseline (static-unit proxy)',
)
ax.plot(
    radii_kpc,
    v_fixed_km_s,
    'g-.',
    linewidth=2.5,
    label='Rigorous GR-style temporal mapping: $r\\ddot{u}_m$ residue',
)
ax.plot(
    radii_kpc,
    v_gm_km_s,
    'b-',
    linewidth=3,
    alpha=0.4,
    label='General Mechanics (2D confinement): pure $1/r$ frame',
)

ax.set_title("Corollary VI: Galactic Rotation (GM update + GR comparison)", fontsize=13)
ax.set_xlabel("Radius from Galactic Center (kpc)")
ax.set_ylabel("Orbital Speed (km/s)")
ax.set_ylim(0, 300)
ax.legend(loc='lower right', fontsize=10)
ax.grid(True, linestyle=':', alpha=0.7)

plt.tight_layout()
plt.show()

# Print text outputs matching prior tests
print("\n[Test VI] Galactic Rotation (r >> R_core)")
print(f"  Observational target       : {v_catalog_km_s:.1f} km/s (Flat)")
print(f"  Classical at 24 kpc        : {v_classical_km_s[-1]:.1f} km/s (Falls below flat target)")
print(f"  Rigorous GR-style at 24 kpc: {v_fixed_km_s[-1]:.1f} km/s (Matches target)")
print(f"  GM 2D confinement at 24 kpc: {v_gm_km_s[-1]:.1f} km/s (Matches target)")
```
